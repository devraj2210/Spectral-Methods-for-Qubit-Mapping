```python
import networkx as nx
from helper_functions import maxcut_results, plot_histogram

import warnings
warnings.filterwarnings("ignore")

from qiskit_ibm_runtime import QiskitRuntimeService

service = QiskitRuntimeService()
backend = service.backend()

print(backend.basis_gates)
#print(backend.configuration().basis_gates)
```

```python
# change gate according to backend 
gate='cz'
```

# **RR4**

```python
rr4_40 = nx.random_regular_graph(n=40, d=4, seed=0)
print('random cut size: ', 0.5 * rr4_40.number_of_edges())
```

### DEFAULT

```python
rr4_40_def = maxcut_results(sim=backend, 
                            backend=backend, 
                            graph_type='reg', 
                            num_nodes=40, 
                            param=4, 
                            default=True, 
                            gate=gate, 
                            average_over=5)
```

```python
print(f"Mean: {rr4_40_def[1]}, Std Dev: {rr4_40_def[2]}")

# might need to reduce y_lim to see the histogram better for 40 qubits
plot_histogram(rr4_40_def[0], keep_top=None, y_lim=0.015)
```

### FOM

```python
rr4_40_fom = maxcut_results(sim=backend, 
                            backend=backend, 
                            graph_type='reg', 
                            num_nodes=40, 
                            param=4, 
                            default=False,
                            tech='1D',
                            gate=gate, 
                            average_over=5)
```

```python
print(f"Mean: {rr4_40_fom[1]}, Std Dev: {rr4_40_fom[2]}")
plot_histogram(rr4_40_fom[0], keep_top=None, y_lim=0.015)
```

### SAMk
(change k accordingly)

```python
k=2
rr4_40_samk = maxcut_results(sim=backend, 
                            backend=backend, 
                            graph_type='reg', 
                            num_nodes=40, 
                            param=4, 
                            default=False,
                            tech='2D',
                            k=k,
                            gate=gate, 
                            average_over=5)
```

```python
print(f"Mean: {rr4_40_samk[1]}, Std Dev: {rr4_40_samk[2]}")
plot_histogram(rr4_40_samk[0], keep_top=None, y_lim=0.015)
```
